```python
# Q1) Find out average, min, max, std deviation of 'annual_medical_cost' and 'annual_premium' columns.

def basic_stats(column_name):
    avg = df[column_name].mean()
    min_val = df[column_name].min()
    max_val = df[column_name].max()
    std_dev = df[column_name].std()
    return {
        'average': avg,
        'min': min_val,
        'max': max_val,
        'std_deviation': std_dev
    }

# Q1) Find out average, min, max, std deviation of 'annual_medical_cost' and 'annual_premium' columns.
annual_medical_cost_stats = basic_stats('annual_medical_cost')
annual_premium_stats = basic_stats('annual_premium')
print("Annual Medical Cost Stats:", annual_medical_cost_stats)
print("Annual Premium Stats:", annual_premium_stats)
```
