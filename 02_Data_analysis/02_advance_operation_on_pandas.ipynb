```python
import pandas as pd

data = {
    "product_category": ["Electronics", "Clothing", "Groceries", "Electronics", "Clothing"],
    "sales_amount": [1200, 800, 150, 3000, 600],
}

df = pd.DataFrame(data)

df.head()
```

```python
df.groupby("product_category").agg({"sales_amount": ["sum", "mean", "max"]})

```

```python
# You can pass more then two columns also for group by 

data = {
    "product_category": ["Electronics", "Clothing", "Groceries", "Electronics", "Clothing", "Groceries"],
    "region": ["North", "South", "East", "West", "North", "South"],
    "sales_amount": [1200, 800, 150, 3000, 600, 400],
}

df = pd.DataFrame(data)

df.head()
```

```python
df.groupby(["region", "product_category"]).agg({"sales_amount": ["sum", "mean", "max"]})
# this is very similar to sql group by operation 
# select region, product_category, sum(sales_amount), avg(sales_amount), max(sales_amount) from table group by region, product_category
# you can also compare it with excel pivot table operation
```

```python
sales_report = df.groupby(["region", "product_category"]).agg({"sales_amount": ["sum", "mean", "max"]})
sales_report.to_excel("sales_report.xlsx")  # Exporting the report to an Excel file
```

```python
# sometime we need to join different dataframe to get the desired result. For example, we have two dataframes, one with product details and another with sales details. We can join them based on a common column like 'product_id' to get a comprehensive view of sales by product category.
product_data = {
    "product_id": [1, 2, 3],
    "product_category": ["Electronics", "Clothing", "Groceries"],
}
sales_data = {
    "product_id": [1, 2, 1, 3, 8],
    "sales_amount": [1200, 800, 3000, 150, 600],
}

df_products = pd.DataFrame(product_data)
df_sales = pd.DataFrame(sales_data)
```

```python
df_products.head()
```

```python
df_sales.head()
```

```python
#lets join : left join
amount = pd.merge(df_sales, df_products, on="product_id", how="left")
```

```python
# lets create one more dataframe where product_id , region and profit_margin is given
region_data = {
    "product_id": [1, 2, 3],
    "region": ["North", "South", "East"],
    "profit_margin": [0.2, 0.15, 0.1],
}
df_regions = pd.DataFrame(region_data)
```

```python
final_df = pd.merge(amount, df_regions, on="product_id", how="left")
```

```python
final_df.head(2)

# lets create more coumns based on existing columns
final_df["total_profit"] = final_df["sales_amount"] * final_df["profit_margin"]
final_df.head(2)
```
